```python
from pathlib import Path
import json

TRAINING_READY_FILE = Path("p4_ds_training_ready.jsonl")
START = 0
END = 200

def preview_dataset(path: Path, start: int, end: int):
    with open(path, "r", encoding="utf-8") as f:
        for i, line in enumerate(f):
            if start <= i <= end:
                entry = json.loads(line)
                print(f"\n🔹 Entry {i + 1}")
                # print(f"🗂 Repo: {entry.get('repo_name', 'N/A')}")
                # print(f"📜 License: {entry.get('license')}, Inferred from: {entry.get("inferred_from", 'N/A')}")
                # print(f"🧬 Version: {entry.get('p4_version')}")
                # print(f"📄 Path: {entry.get('p4_file_path')}")
                print(f" Description: {entry.get('desc')}")
                # print(f"🧹 Cleaned code:\n{entry.get('cleaned_p4', '').strip()[0:150]} \n ... \n")  # Truncated for preview
                # print("-" * 80)
            
            if i > end:
                break

if __name__ == "__main__":
    preview_dataset(TRAINING_READY_FILE, START, END)
```
